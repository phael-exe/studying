# Implementação do Harris Corner Detector

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
%matplotlib inline

image = cv2.imread('img/gengarjdm.jpg')

image_copy = np.copy(image)

image_copy = cv2.cvtColor(image_copy, cv2.COLOR_BGR2RGB)

plt.imshow(image_copy)
```

```python
gray = cv2.cvtColor(image_copy, cv2.COLOR_RGB2GRAY)
gray = np.float32(gray)

dst = cv2.cornerHarris(gray, 2, 3, 0.04)

dst = cv2.dilate(dst, None)

plt.imshow(dst, cmap='gray')
```

```python
thresh1 = 0.1*dst.max()
thresh2 = 0.04*dst.max()
thresh3 = 0.2*dst.max()
```

```python
corner_image = np.copy(image_copy)
```

```python
def threshcomparator(thresh):
    corner_image = np.copy(image_copy)  # reseta a imagem a cada chamada
    for j in range(0, dst.shape[0]):
        for i in range(0, dst.shape[1]):
            if(dst[j, i] > thresh):
                cv2.circle(corner_image, (i,j), 1, (0, 255, 0), 1)

    return plt.imshow(corner_image)
```

```python
threshcomparator(thresh1)
```

```python
threshcomparator(thresh2)
```

```python
threshcomparator(thresh3)
```
