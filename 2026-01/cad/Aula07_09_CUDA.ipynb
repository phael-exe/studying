# Evitando acessos fora da faixa do vetor

```python
%%cuda
#include <stdio.h>


__global__ void initializeElementsTo(int initialValue, int *a, int N)
{
  int i = threadIdx.x + blockIdx.x * blockDim.x;
  if (i < N)
  {
    a[i] = initialValue;
  }
}

int main()
{
  /*
   * Do not modify `N`.
   */

  int N = 1000;

  int *a;
  size_t size = N * sizeof(int);

  cudaMallocManaged(&a, size);

  /*
   * Assume we have reason to want the number of threads
   * fixed at `256`: do not modify `threads_per_block`.
   */

  size_t threads_per_block = 256;

  /*
   * The following is idiomatic CUDA to make sure there are at
   * least as many threads in the grid as there are `N` elements.
   */

  size_t number_of_blocks = 4;
  //size_t number_of_blocks = (N + threads_per_block - 1) / threads_per_block;

  int initialValue = 6;

  initializeElementsTo<<<number_of_blocks, threads_per_block>>>(initialValue, a, N);
  cudaDeviceSynchronize();

  /*
   * Check to make sure all values in `a`, were initialized.
   */

  for (int i = 0; i < N; ++i)
  {
    if(a[i] != initialValue)
    {
      printf("FAILURE: target value: %d\t a[%d]: %d\n", initialValue, i, a[i]);
      cudaFree(a);
      exit(1);
    }
  }
  printf("SUCCESS!\n");

  cudaFree(a);
}
```

# Questão 2

Considerando os exemplos acima, crie um vetor de 100 posições e inicialize-o de forma parelela com valor 100. Em seguida, incremente-o em ordem crescente (de 0 à 99) em cada posição do vetor (da posição 0 à 99). Confira se seu resultado está correto.

```python
%%cuda
#include <stdio.h>

__global__ void inicializar(int *a, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    if (i < N) a[i] = 100;
}

__global__ void incrementar(int *a, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    if (i < N) a[i] += i;
}

int main() {
    int N = 100;
    int *a;
    cudaMallocManaged(&a, N * sizeof(int));

    inicializar<<<4, 32>>>(a, N);
    cudaDeviceSynchronize();

    incrementar<<<4, 32>>>(a, N);
    cudaDeviceSynchronize();

    int ok = 1;
    for (int i = 0; i < N; i++) {
        if (a[i] != 100 + i) { ok = 0; break; }
    }
    printf(ok ? "SUCCESS!\n" : "FAILURE!\n");

    cudaFree(a);
    return 0;
}
```

# Questão 3

 Considerando os exemplos acima, crie dois vetores de 1000 posições e inicialize o primeiro vetor de forma parelela com valor decrescente (de 0 à 999) e o segundo vetor com valores crescentes (de 0 à 999). Em seguida, faça a soma dos dois vetores, elemento a elemento. Crie uma função sequencial para retornar se todos os valores do vetor resultante são iguais.

```python
%%cuda
#include <stdio.h>

__global__ void init_decrescente(int *a, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    int stride = blockDim.x * gridDim.x;
    for (; i < N; i += stride) a[i] = (N - 1) - i;
}

__global__ void init_crescente(int *b, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    int stride = blockDim.x * gridDim.x;
    for (; i < N; i += stride) b[i] = i;
}

__global__ void somar(int *a, int *b, int *c, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    int stride = blockDim.x * gridDim.x;
    for (; i < N; i += stride) c[i] = a[i] + b[i];
}

void verificar(int *c, int N) {
    for (int i = 0; i < N; i++) {
        if (c[i] != c[0]) {
            printf("FAILURE: c[%d] = %d (esperado %d)\n", i, c[i], c[0]);
            return;
        }
    }
    printf("SUCCESS! Todos os valores sao iguais a %d.\n", c[0]);
}

int main() {
    int N = 1000;
    int *a, *b, *c;
    cudaMallocManaged(&a, N * sizeof(int));
    cudaMallocManaged(&b, N * sizeof(int));
    cudaMallocManaged(&c, N * sizeof(int));

    init_decrescente<<<8, 16>>>(a, N);
    init_crescente<<<8, 16>>>(b, N);
    cudaDeviceSynchronize();

    somar<<<8, 16>>>(a, b, c, N);
    cudaDeviceSynchronize();

    verificar(c, N);

    cudaFree(a); cudaFree(b); cudaFree(c);
    return 0;
}
```

#Questão 4


Implemente um código para inicializar dois vetores de tamanho N ($N=70$). O primeiro vetor deve ser inicializado com valor igual a 10 e o segundo vetor deve ser inicializado com valor crescente (de 0 a $N-1$). Imprima a multiplicação do primeiro vetor com o segundo vetor, elemento a elemento. Obs.: Considere o bloco com 32 threads.

```python
%%cuda
#include <stdio.h>

__global__ void init_constante(int *v, int val, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    if (i < N) v[i] = val;
}

__global__ void init_crescente(int *v, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    if (i < N) v[i] = i;
}

__global__ void multiplicar(int *v1, int *v2, int *res, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    if (i < N) res[i] = v1[i] * v2[i];
}

int main() {
    int N = 70;
    int *v1, *v2, *res;
    cudaMallocManaged(&v1,  N * sizeof(int));
    cudaMallocManaged(&v2,  N * sizeof(int));
    cudaMallocManaged(&res, N * sizeof(int));

    int blocos = (N + 31) / 32;  // ceil(70/32) = 3

    init_constante<<<blocos, 32>>>(v1, 10, N);
    init_crescente<<<blocos, 32>>>(v2, N);
    cudaDeviceSynchronize();

    multiplicar<<<blocos, 32>>>(v1, v2, res, N);
    cudaDeviceSynchronize();

    for (int i = 0; i < N; i++)
        printf("res[%d] = %d\n", i, res[i]);

    cudaFree(v1); cudaFree(v2); cudaFree(res);
    return 0;
}
```

#Questão 5

```python
%%cuda
#include <stdio.h>

__global__ void init_crescente(int *a, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    int stride = blockDim.x * gridDim.x;
    for (; i < N; i += stride) a[i] = i;
}

__global__ void elevar_quadrado(int *a, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    int stride = blockDim.x * gridDim.x;
    for (; i < N; i += stride) a[i] = a[i] * a[i];
}

int main() {
    int N = 1000;
    int *a;
    cudaMallocManaged(&a, N * sizeof(int));

    init_crescente<<<32, 256>>>(a, N);
    cudaDeviceSynchronize();

    elevar_quadrado<<<32, 256>>>(a, N);
    cudaDeviceSynchronize();

    int ok = 1;
    for (int i = 0; i < N; i++) {
        if (a[i] != i * i) { ok = 0; printf("FAILURE: a[%d] = %d (esperado %d)\n", i, a[i], i*i); break; }
    }
    if (ok) printf("SUCCESS!\n");

    cudaFree(a);
    return 0;
}
```

#Questão 6

Proponha uma solução paralela em CUDA para o cálculo da seguinte expressão vetorial: $V=k_1*A+k_2*B$, onde $k_1$ e $k_2$ são constantes e $A$ e $B$ são vetores de tamanho $n=500$. Utilize $n_{blocos}=8$ e $n_{threads}=16$. Inicialize os vetores $A$ e $B$ de forma paralela e com valores arbitrários (definidos pelo programador). Imprima o vetor $V$.

```python
%%cuda
#include <stdio.h>

__global__ void init_A(float *A, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    int stride = blockDim.x * gridDim.x;
    for (; i < N; i += stride) A[i] = (float)i * 2.0f;  // valores arbitrários
}

__global__ void init_B(float *B, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    int stride = blockDim.x * gridDim.x;
    for (; i < N; i += stride) B[i] = (float)(N - i);  // valores arbitrários
}

__global__ void calcular_V(float *A, float *B, float *V, float k1, float k2, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    int stride = blockDim.x * gridDim.x;
    for (; i < N; i += stride) V[i] = k1 * A[i] + k2 * B[i];
}

int main() {
    int N = 500;
    float k1 = 2.0f, k2 = 3.0f;
    float *A, *B, *V;

    cudaMallocManaged(&A, N * sizeof(float));
    cudaMallocManaged(&B, N * sizeof(float));
    cudaMallocManaged(&V, N * sizeof(float));

    init_A<<<8, 16>>>(A, N);
    init_B<<<8, 16>>>(B, N);
    cudaDeviceSynchronize();

    calcular_V<<<8, 16>>>(A, B, V, k1, k2, N);
    cudaDeviceSynchronize();

    for (int i = 0; i < N; i++)
        printf("V[%d] = %.2f\n", i, V[i]);

    cudaFree(A); cudaFree(B); cudaFree(V);
    return 0;
}
```

Com pré-busca (prefetch)

```python
%%cuda
#include <stdio.h>

__global__ void init_crescente(int *a, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    int stride = blockDim.x * gridDim.x;
    for (; i < N; i += stride) a[i] = i;
}

__global__ void init_decrescente(int *b, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    int stride = blockDim.x * gridDim.x;
    for (; i < N; i += stride) b[i] = (N - 1) - i;
}

__global__ void somar(int *a, int *b, int *c, int N) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    int stride = blockDim.x * gridDim.x;
    for (; i < N; i += stride) c[i] = a[i] + b[i];
}

void verificacao(int *c, int N) {
    for (int i = 0; i < N; i++) {
        if (c[i] != c[0]) {
            printf("Erro detectado no codigo.\n");
            exit(1);
        }
    }
    printf("Sucesso! Todos os valores sao identicos.\n");
}

int main() {
    int N = 1000;
    int device;
    cudaGetDevice(&device);

    int *a, *b, *c;
    cudaMallocManaged(&a, N * sizeof(int));
    cudaMallocManaged(&b, N * sizeof(int));
    cudaMallocManaged(&c, N * sizeof(int));

    // Pré-busca para a GPU antes dos kernels
    cudaMemPrefetchAsync(a, N * sizeof(int), device, NULL);
    cudaMemPrefetchAsync(b, N * sizeof(int), device, NULL);
    cudaMemPrefetchAsync(c, N * sizeof(int), device, NULL);

    init_crescente<<<8, 16>>>(a, N);
    init_decrescente<<<8, 16>>>(b, N);
    cudaDeviceSynchronize();

    somar<<<8, 16>>>(a, b, c, N);
    cudaDeviceSynchronize();

    // Pré-busca de volta para a CPU antes da verificação
    cudaMemPrefetchAsync(c, N * sizeof(int), cudaCpuDeviceId, NULL);
    cudaDeviceSynchronize();

    verificacao(c, N);

    cudaFree(a); cudaFree(b); cudaFree(c);
    return 0;
}
```

Analise o código abaixo e o paralelize:

```python
#include <stdio.h>

// Função para somar os elementos de um vetor sequencialmente
int somaSequencial(int* vetor, int tamanho) {
    int soma = 0;
    for (int i = 0; i < tamanho; i++) {
        soma += vetor[i];
    }
    return soma;
}

int main() {
    const int tamanho = 100000; // Tamanho do vetor
    int vetor[tamanho];

    // Inicializa o vetor com valores aleatórios (simulação)
    for (int i = 0; i < tamanho; i++) {
        vetor[i] = i;
    }

    // Chama a função de soma sequencial
    int resultado = somaSequencial(vetor, tamanho);

    printf("Resultado da soma sequencial: %d\n", resultado);

    return 0;
}
```

```python
%%cuda
#include <stdio.h>

__global__ void somaParalela(int *vetor, long long *resultado, int tamanho) {
    int i = threadIdx.x + blockIdx.x * blockDim.x;
    int stride = blockDim.x * gridDim.x;
    for (; i < tamanho; i += stride) {
        atomicAdd(resultado, (long long)vetor[i]);
    }
}

int main() {
    const int tamanho = 100000;
    int *vetor;
    long long *resultado;

    cudaMallocManaged(&vetor,    tamanho * sizeof(int));
    cudaMallocManaged(&resultado, sizeof(long long));
    *resultado = 0;

    for (int i = 0; i < tamanho; i++) vetor[i] = i;

    int blocos = 256, threads = 256;
    somaParalela<<<blocos, threads>>>(vetor, resultado, tamanho);
    cudaDeviceSynchronize();

    printf("Resultado da soma paralela: %lld\n", *resultado);
    // Valor esperado: 0+1+...+99999 = 4999950000

    cudaFree(vetor);
    cudaFree(resultado);
    return 0;
}
```
